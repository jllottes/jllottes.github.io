Last week, we started to look at working with text data. In particular, we loaded in the text from *Frankenstein; or, The Modern Prometheus*. 

```python
with open('frankenstein.txt', 'r', encoding='utf-8-sig') as f:  # Opens the file for reading
    text = f.read()                                            # Reads the content into a string
```

```python
word_count_dict = {}

words = text.split()
for word in words:
    if not word in word_count_dict:   # If we've not previously seen word
        word_count_dict[word] = 1     # add it to the dictionary with count 1
    else:                             # Otherwise, if we have seen the word previously
        word_count_dict[word] += 1    # then we bump up the count by 1
```

```python
print(words[:50])
```

**Exercise:** Use the `.replace` method to remove punctuation marks and other non-letter characters from the text of *Frankenstein*.
```python
cleaned_text = text.lower()

#chars_to_remove = [',', '.', '!', '?', '"', "'", ...]
chars_to_remove = ',.!?()"#@%&*:;' + "'"

for char in chars_to_remove:
    cleaned_text = cleaned_text.replace(char, '')    # Remove char from the string

words = cleaned_text.split()
```

```python
print(words[:50])
```

```python
word_count_dict = {}
for word in words:
    if not word in word_count_dict:   # If we've not previously seen word
        word_count_dict[word] = 1     # add it to the dictionary with count 1
    else:                             # Otherwise, if we have seen the word previously
        word_count_dict[word] += 1    # then we bump up the count by 1
```

### Background:  ASCII codes

Each character on a computer keyboard is assigned an [ASCII code](http://www.theasciicode.com.ar/), which is an integer in the range `0`-`127`. The ASCII code of a character can be obtained using the `ord()` function:

```python
for c in "This is MTH 337":
    print("'{}'  ->  {}".format(c, ord(c)))
```

Conversely, the function `chr()` converts ASCII codes into characters:

```python
ascii_list = [104, 101, 108, 108, 111]

char_list = []
for n in ascii_list:
    char_list.append(chr(n))
    
txt = ''.join(char_list)
print(txt)
```
